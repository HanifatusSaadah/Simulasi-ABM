```python
# =========================
# SIMULASI PERILAKU TY1
# MUSIM KEMARAU (6 BULAN)
# =========================

# ------- DATA AWAL -------
singkong = 22
talas = 18
ubi_jalar = 16
tomat = 5
beluntas = 7
bayam = 12
kemangi = 3
cabai = 8
lengkuas = 13
jinten = 2
pandan = 6
kacang_tanah = 20
pisang = 5
pepaya = 4
mangga = 1
sirsak = 1
kopi = 1

T = singkong + talas + ubi_jalar + tomat + beluntas + bayam + kemangi + cabai + lengkuas + jinten + pandan + kacang_tanah + pisang + pepaya + mangga + sirsak + kopi

# ------ PROBABILITAS ------
P_tambah = 0.50
P_kurang = 0.33

# ------ PROPORSI AWAL ------
P_singkong = singkong / T
P_talas = talas / T
P_ubi_jalar = ubi_jalar / T
P_tomat = tomat / T
P_beluntas = beluntas / T
P_bayam = bayam / T
P_kemangi = kemangi / T
P_cabai = cabai / T
P_lengkuas = lengkuas / T
P_jinten = jinten / T
P_pandan = pandan / T
P_kacang_tanah = kacang_tanah / T
P_pisang = pisang / T
P_pepaya = pepaya / T
P_mangga = mangga / T
P_sirsak = sirsak / T
P_kopi = kopi / T
# ------ SIMULASI 6 BULAN ------
bulan = 6
hasil_T = [T]

for t in range(bulan):

   A = P_tambah * T
   R = P_kurang * T

   T = T + (A - R)

   hasil_T.append(T)

# ------ HASIL TOTAL ------
print("Total tanaman tiap bulan:")
for i, val in enumerate(hasil_T):
    print(f"Bulan {i}: {val:.2f}")

# ------ HASIL AKHIR PER JENIS ------
T_akhir = hasil_T[-1]

singkong_akhir = P_singkong * T_akhir
talas_akhir = P_talas * T_akhir
ubi_jalar_akhir = P_ubi_jalar * T_akhir
tomat_akhir = P_tomat * T_akhir
beluntas_akhir = P_beluntas * T_akhir
bayam_akhir = P_bayam * T_akhir
kemangi_akhir = P_kemangi * T_akhir
cabai_akhir = P_cabai * T_akhir
lengkuas_akhir = P_lengkuas * T_akhir
jinten_akhir = P_jinten * T_akhir
pandan_akhir = P_pandan * T_akhir
kacang_tanah_akhir = P_kacang_tanah * T_akhir
pisang_akhir = P_pisang * T_akhir
pepaya_akhir = P_pepaya * T_akhir
mangga_akhir = P_mangga * T_akhir
sirsak_akhir = P_sirsak * T_akhir
kopi_akhir = P_kopi * T_akhir

print("\nJumlah tanaman akhir:")
print("singkong:", round(singkong_akhir, 2))
print("talas:", round(talas_akhir, 2))
print("ubi jalar:", round(ubi_jalar_akhir, 2))
print("tomat:", round(tomat_akhir, 2))
print("beluntas:", round(beluntas_akhir, 2))
print("bayam:", round(bayam_akhir, 2))
print("kemangi:", round(kemangi_akhir, 2))
print("cabai:", round(cabai_akhir, 2))
print("lengkuas:", round(lengkuas_akhir, 2))
print("jinten:", round(jinten_akhir, 2))
print("pandan:", round(pandan_akhir, 2))
print("kacang tanah:", round(kacang_tanah_akhir, 2))
print("pisang:", round(pisang_akhir, 2))
print("pepaya:", round(pepaya_akhir, 2))
print("mangga:", round(mangga_akhir, 2))
print("sirsak:", round(sirsak_akhir, 2))
print("kopi:", round(kopi_akhir, 2))
```
